```python
import httpx
from langchain_ollama import ChatOllama

# Check if Ollama is running
try:
    response = httpx.get("http://localhost:11434/api/tags", timeout=2.0)
    print("✓ Ollama is running")
except (httpx.ConnectError, httpx.ConnectTimeout):
    print("✗ ERROR: Ollama is not running!")
    print("\nTo start Ollama, run one of these commands:")
    print("  - If installed via package manager: systemctl start ollama")
    print("  - If installed manually: ollama serve")
    print("  - Or check: https://ollama.ai/download")
    raise

# Initialize the LLM
llm = ChatOllama(model="deepseek-r1:8b")
print(f"✓ Initialized ChatOllama with model: deepseek-r1:8b")
```

```python
try:
    print("Question: What is the 3rd root of 27?")
    print("Answer: ", end="")
    for chunk in llm.stream("What is the 3rd root of 27?"):
        print(chunk.content, end="")
    print()  # New line after streaming
except Exception as e:
    print(f"\n✗ Error during streaming: {e}")
    print("\nPossible issues:")
    print("  1. Ollama service is not running")
    print("  2. Model 'deepseek-r1:8b' is not pulled/downloaded")
    print("     Run: ollama pull deepseek-r1:8b")
    raise
```
